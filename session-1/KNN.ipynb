```python
#KNN
import numpy as np
```

```python
data = [
    [150, 7.0, 1, 'Apple'],
    [120, 6.5, 0, 'Banana'],
    [180, 7.5, 2, 'Orange'],
    [155, 7.2, 1, 'Apple'],
    [110, 6.0, 0, 'Banana'],
    [190, 7.8, 2, 'Orange'],
    [145, 7.1, 1, 'Apple'],
    [115, 6.3, 0, 'Banana']
]
data_array=np.array(data)
label={'Apple':0,'Banana':1,'Orange':2}
X=data_array[:,0:3].astype(float)
Y=np.array([label[item] for item in data_array[:, 3]])
```

```python
def euclidean_distance(p1,p2):
    return np.sqrt(np.sum((p1 -p2)**2))
```

```python
class KNN:
    def __init__(self,k=3):
        self.k=k
    def fit(self,X,Y):
        self.X_train=X
        self.Y_train=Y
    def predict_one(self,x):
        distances=[euclidean_distance(x,x_train) for x_train in self.X_train]
        k_indices=np.argsort(distances)[:self.k]
        k_nearest_labels=self.Y_train[k_indices]
        most_common=np.bincount(k_nearest_labels.astype(int)).argmax()
        return most_common
    def predict(self, X_test):
        return np.array([self.predict_one(x) for x in X_test])   
```

```python
test_data = np.array([
    [118, 6.2, 0],  # Expected: Banana
    [160, 7.3, 1],  # Expected: Apple
    [185, 7.7, 2]   # Expected: Orange
])
model=KNN(k=3)
model.fit(X,Y)
predicted=model.predict(test_data)
print(predicted)
```
